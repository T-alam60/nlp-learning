# chat word tyreatment 

```python
chat_words = {
    'u': 'you',
    'ur': 'your',
    'r': 'are',
    'btw': 'by the way',
    'idk': 'i dont know',
    'omg': 'oh my god',
    'brb': 'be right back',
    'lol': 'laughing out loud',
    'thx': 'thanks',
    'pls': 'please'
}

def chat_word_treatment(text):
    words = text.split()
    
    new_words = []
    
    for word in words:
        new_words.append(chat_words.get(word.lower(), word))
    
    return ' '.join(new_words)
```

```python
text = "u r gr8 btw pls call me"

print(chat_word_treatment(text))
```

# removing stop word 

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    
    filtered_words = [
        word for word in words 
        if word.lower() not in stop_words
    ]
    
    return ' '.join(filtered_words)
```

```python
df['review'] = df['review'].apply(remove_stopwords)
```

# tokenization 


```python
# word 
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')

def tokenize_text(text):
    return word_tokenize(text)
```

```python
text = "I really love this movie."

print(tokenize_text(text))
```

```python
# text 
from nltk.tokenize import sent_tokenize

def sentence_tokenize(text):
    return sent_tokenize(text)
```

```python
text = "I love this movie. The acting was amazing."

print(sentence_tokenize(text))
```

# stemming 

```python
import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

def stemming(text):
    words = text.split()
    
    stemmed_words = [stemmer.stem(word) for word in words]
    
    return ' '.join(stemmed_words)
```

```python
text = "I am playing cricket and he played yesterday"

print(stemming(text))
```

# Lemmatization 

```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

def lemmatization(text):
    words = text.split()

    lemmatized_words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]

    return ' '.join(lemmatized_words)
```

```python
text = "The cats are running and the children are playing"

print(lemmatization(text))
```

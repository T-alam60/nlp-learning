```python
import pandas as pd

data = {
    'Document': ['D1', 'D2', 'D3', 'D4'],
    'Text': [
        'people watch campus',
        'campusx watch campusx',
        'people write comment',
        'campusx write comment'
    ],
    'Target': [1, 1, 0, 0]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer()
```

```python
bow = cv.fit_transform(df['Text'])
```

```python
# vocab
print(cv.vocabulary_)
```

```python
print(bow[0].toarray())
print(bow[1].toarray())
```

```python
# it handle out of vocalbre 
cv.transform(["campusx wath and write comment of campusx"]).toarray()
```
